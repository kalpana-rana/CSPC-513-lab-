Aim:write a python program to import and export data using pandas library function.

Import data : bring external data to python.
use pandas to import : pd.read_csv() , pd.read_excel(),pd.read_json().

We can perform some operations on this data using pandas.


Export data : take data(updated) from python (dataframe) and save it somewhere or to the same file.
use pandas to export : df.to_csv() , df.to_excel() , df.to_json()

Note- When you export the data from datframe to csv(or any) then pandas automatically include the row numbers as the very first column which result into unnamed columns , if you read and write file repeatdly thne pandas will add new index column everytime. So to prevent this behaviour use index=False. 

                                                             importing and exporting csv file        

```python

#IMPORT:
import pandas as pd
df = pd.read_csv("datasets/marvel_ratings.csv") #importing dataset.
df.head(5) #importing only first 5 rows.
df #importing whole dataset.
```

```python
#EXPORT
import pandas as pd
df = pd.read_csv("datasets/marvel_ratings.csv") #importing dataset.
df.drop_duplicates() #removing all duplicate records.
df.to_csv("datasets/marvel_ratings_export.csv" , index = False)
```

                                                         importing and exporting excel file

```python
#IMPORT:
import pandas as pd
df = pd.read_excel("datasets/marvel_ratings.xls") #importing dataset.
df.head()
```

```python
#EXPORT:
import pandas as pd
df = pd.read_excel("datasets/marvel_ratings.xls")
df.info() #summary of dataframe
df.to_excel("/Users/kalpanarana/Downloads/Cola dataset in xlsx.xlsx" , index = False)
```

                                                     importing and exporting json file

```python
#IMPORT:
import pandas as pd
df = pd.read_json("datasets/marvel_ratings.json") #importing dataset.
df
```

```python
#EXPORT:
import pandas as pd
df = pd.read_json("datasets/marvel_ratings.json")
df.value_counts() #checking the no. of missing values in each record.
df.to_json("datasets/marvel_ratings.json" , index = False)
```
